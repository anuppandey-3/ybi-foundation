# Basics of Logistic Regression Model

### What is Logistic Regression?

**Logistic Regression** is a fundamental statistical and machine learning algorithm used for **classification** tasks. Unlike Linear Regression, which predicts continuous numerical values, Logistic Regression predicts the probability of an event occurring (e.g., whether an email is spam or not, or if a transaction is fraudulent).

### Key Concepts

#### 1. Binary vs. Multinomial Classification
* **Binary Classification (most common):** The target variable has only two possible outcomes (e.g., `0` or `1`, `Yes` or `No`).
* **Multinomial Classification:** The target variable has three or more unordered categories (e.g., predicting if an image is a cat, dog, or bird).

#### 2. The Sigmoid (Logistic) Function
Linear regression outputs values from negative infinity to positive infinity. Logistic regression maps these real-valued predictions to a probability between `0` and `1` using the **Sigmoid Function**:

$$S(z) = \frac{1}{1 + e^{-z}}$$

Where:
* $z$ is the input to the function (the linear combination of features: $z = \beta_0 + \beta_1x_1 + ... + \beta_nx_n$).
* $S(z)$ represents the probability (ranging from 0 to 1).

#### 3. Decision Boundary
To make a final class prediction (e.g., Class 0 or Class 1), a threshold is set (commonly `0.5`):
* If $P(Y=1) \ge 0.5$, predict **Class 1**.
* If $P(Y=1) < 0.5$, predict **Class 0**.

#### 4. Cost Function: Log Loss
Instead of Mean Squared Error (used in linear regression), logistic regression uses **Log Loss** (or Binary Cross-Entropy) because the sigmoid function introduces non-linearity, which would make MSE non-convex (hard to optimize):

$$\text{Cost}(h_\theta(x), y) = -y \log(h_\theta(x)) - (1 - y) \log(1 - h_\theta(x))$$
