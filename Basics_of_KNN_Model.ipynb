# Basics of KNN Model

### What is K-Nearest Neighbors (KNN)?

**K-Nearest Neighbors (KNN)** is a simple, intuitive, and versatile **supervised machine learning algorithm** used for both **classification** and **regression** tasks. It is often referred to as a **lazy learner** or **instance-based learner** because it does not explicitly build a model during a training phase. Instead, it memorizes the training dataset and performs computations only when making predictions for new data points.

---

### How KNN Works (Step-by-Step)

1. **Choose the number of neighbors, $K$:** Decide how many historical data points you want to compare your new data point against.
2. **Calculate the distance:** Compute the distance (e.g., Euclidean distance) between the new, unseen data point and all the points in the training dataset.
3. **Find the nearest neighbors:** Sort the distance values in ascending order and select the top $K$ closest training data points.
4. **Make a prediction:**
   * **For Classification:** Use a **majority vote**. The new data point is assigned to the class that is most common among its $K$ nearest neighbors.
   * **For Regression:** Calculate the **average (mean)** value of the target variables of the $K$ nearest neighbors to predict the value for the new data point.

---

### Core Concepts & Components

#### 1. The Importance of $K$
* **Small $K$ (e.g., $K=1$):** The model is highly sensitive to noise in the data, which can lead to **overfitting**.
* **Large $K$:** The model becomes smoother and more stable, but if $K$ is too large, it can include points from other classes/groups, leading to **underfitting**.
* **Choosing $K$:** A common heuristic is to use $K = \sqrt{N}$ (where $N$ is the total number of training samples) or use **Cross-Validation** to find the optimal $K$. For binary classification, choosing an **odd number** for $K$ is recommended to avoid tie votes.

#### 2. Distance Metrics
The performance of a KNN model heavily relies on how "closeness" is measured. Common distance metrics include:
* **Euclidean Distance** (most common for continuous variables): Calculates the straight-line distance between two points.
  $$d(p, q) = \sqrt{\sum_{i=1}^{n} (p_i - q_i)^2}$$
* **Manhattan Distance** (useful for grid-based paths or high-dimensional spaces):
  $$d(p, q) = \sum_{i=1}^{n} |p_i - q_i|$$
* **Cosine Similarity** (often used for text classification or recommendation systems based on orientation rather than magnitude).
