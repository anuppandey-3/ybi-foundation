# Python Lists

```markdown
## Creating Lists
Lists in Python are created by placing all the items (elements) inside square brackets `[]`, separated by commas.

A list can have any number of items and they may be of different types (integer, float, string, etc.).
```

````python
# Creating an empty list
empty_list = []

# List of integers
numbers = [1, 2, 3, 4, 5]

# List with mixed data types
mixed_list = [1, "Hello", 3.4]

print("Empty:", empty_list)
print("Numbers:", numbers)
print("Mixed:", mixed_list)
````

```markdown
## Accessing Elements
There are various ways in which we can access the elements of a list:

1. **Index**: We can use the index operator `[]` to access an item in a list. In Python, indices start at `0`.
2. **Negative Indexing**: Python allows negative indexing for its sequences. The index of `-1` refers to the last item, `-2` to the second last item, and so on.
3. **Slicing**: We can access a range of items in a list by using the slicing operator `:` (colon).
```

````python
my_list = ['p', 'y', 't', 'h', 'o', 'n']

# Accessing by positive index
print("First element:", my_list[0])
print("Third element:", my_list[2])

# Accessing by negative index
print("Last element:", my_list[-1])
print("Second to last element:", my_list[-2])

# Slicing: elements from index 1 to 4 (exclusive)
print("Slice [1:4]:", my_list[1:4])

# Slicing: from beginning to index 3 (exclusive)
print("Slice [:3]:", my_list[:3])

# Slicing: from index 3 to the end
print("Slice [3:]:", my_list[3:])
````
