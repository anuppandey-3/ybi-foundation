# ANN Classification

### Load Dataset for Classification

We will load the **Wine dataset** from `scikit-learn`, which is a classic multiclass classification dataset. We will convert it into a pandas DataFrame to easily inspect the features and targets.

```python
import pandas as pd
from sklearn.datasets import load_wine

# Load the raw dataset
wine_data = load_wine()

# Convert to a pandas DataFrame
df = pd.DataFrame(data=wine_data.data, columns=wine_data.feature_names)
df['target'] = wine_data.target

# Display target class names and DataFrame head
print(f"Target classes to classify: {list(wine_data.target_names)}")
display(df.head())
```

```python
# Calculate and display summary statistics for the features
summary_stats = df.describe().T
display(summary_stats)
```
