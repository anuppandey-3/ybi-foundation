# Python Tuples

## Creating a Tuple

A tuple is an ordered, immutable collection in Python. You can create them using parentheses `()` or the `tuple()` constructor.

```python
# Creating tuples
my_tuple = (1, 2, 3, 'apple', 'banana')
empty_tuple = ()
single_element_tuple = (42,)  # Note the trailing comma

print("my_tuple:", my_tuple)
print("single_element_tuple:", single_element_tuple)
```

## Accessing Tuple Elements

You can access elements of a tuple using indexing (starting from `0`) or slicing.

```python
# Accessing elements
print("First element:", my_tuple[0])
print("Last element:", my_tuple[-1])
print("Sliced tuple (elements 1 to 3):", my_tuple[1:4])
```
