# Python Sets

### Differences Between Lists and Sets in Python

Both **lists** and **sets** are used to store collections of data in Python, but they have fundamental differences in how they behave and are stored:

| Feature | List (`list`) | Set (`set`) |
| :--- | :--- | :--- |
| **Ordering** | **Ordered** (remains in the order you insert items). | **Unordered** (no guaranteed order). |
| **Duplicates** | **Allowed** (can store identical elements). | **Not Allowed** (every element must be unique). |
| **Indexing & Slicing** | **Supported** (e.g., `my_list[0]`). | **Not Supported** (cannot access via an index). |
| **Mutability** | **Mutable** (can change, add, or remove elements). | **Mutable** (but elements within it must be hashable/immutable). |
| **Performance (Search)** | **Slow** - $O(n)$ time complexity to search. | **Very Fast** - $O(1)$ average time complexity to search. |
| **Syntax** | Defined using square brackets: `[1, 2, 3]`. | Defined using curly braces: `{1, 2, 3}`. |

```python
# 1. Order and Duplicates Demonstration

# Lists preserve order and allow duplicates
my_list = [3, 1, 2, 2, 1]
print("List:", my_list)

# Sets automatically remove duplicates and do not guarantee insertion order
my_set = {3, 1, 2, 2, 1}
print("Set:", my_set)
```

```python
# 2. Indexing Demonstration

try:
    print("First item in list:", my_list[0])
    print("First item in set:", my_set[0])  # This will raise a TypeError
except TypeError as e:
    print("Error accessing set element:", e)
```
