# Python Conditional Statement

### Introduction to `if-else` Statements

In Python, conditional statements allow your program to make decisions and execute different blocks of code based on whether a condition is **True** or **False**.

#### Basic Syntax:
```python
if condition:
    # This block executes if the condition is True
    print("Condition is True")
else:
    # This block executes if the condition is False
    print("Condition is False")
```

#### Key Points:
* **Colons (`:`)**: Every conditional statement line (`if`, `elif`, `else`) must end with a colon.
* **Indentation**: Python uses indentation (usually 4 spaces) to define the block of code inside the condition.
* **`elif` (else if)**: Used to check multiple, sequential conditions if the previous conditions were False.

````python
# Example: Check if a person is eligible to vote
age = 20

if age >= 18:
    print("You are eligible to vote!")
else:
    print("You are too young to vote.")
````
